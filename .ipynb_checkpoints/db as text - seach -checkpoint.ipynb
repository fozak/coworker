```python
import re

# Load the .sql file into memory
with open("erp.sql", "r", encoding="utf-8") as f:
    sql_text = f.read()

# Regex to match anything like ,( ... Jim ... )
pattern = r",\([^)]*Jim V[^)]*\)"

#pattern = r",\(([^()]*?Jim V[^()]*?)\),"

# Find all matches
matches = re.findall(pattern, sql_text)

# Print results
for match in matches:
    print(match)
```
